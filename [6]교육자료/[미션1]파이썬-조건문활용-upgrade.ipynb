## 📌 비교 연산자 총정리

<span style="position:left;display:inline-block;">

| 연산자 | 의미       | 예시          | 결과    |
| ---- | -------- | ----------- | ----- |
| `==` | 같다       | `5 == 5`    | True  |
| `!=` | 다르다      | `5 != 3`    | True  |
| `>`  | 크다       | `7 > 3`     | True  |
| `<`  | 작다       | `2 < 1`     | False |
| `>=` | 크거나 같다   | `5 >= 5`    | True  |
| `<=` | 작거나 같다   | `4 <= 3`    | False |

</span>


````python
## 💻 예제 1 — 나이 판별 프로그램

age = int(input("나이를 입력하세요: "))

if age >= 14:
    print("중학생 이상입니다.")
else:
    print("초등학생입니다.")
````

````python
## 💻 예제 2 — 비밀번호 확인

password = input("비밀번호 입력: ")

if password == "python":
    print("접속 성공!")
else:
    print("비밀번호 오류")

````

````python
## 💻 예제 3 — 홀수 / 짝수 판별기

num = int(input("정수를 입력하세요: "))

if num % 2 == 0:
    print(f"{num}은 짝수입니다.")
else:
    print(f"{num}은 홀수입니다.")

````

````python
## 💻 예제 4 — 온도 판별기

temp = float(input("현재 기온을 입력하세요(°C): "))

if temp >= 30:
    print("☀️ 매우 더운 날씨입니다. 시원하게 입으세요!")
elif temp >= 20:
    print("🌤 따뜻한 날씨입니다. 가볍게 입으세요!")
elif temp >= 10:
    print("🍂 쌀쌀한 날씨입니다. 겉옷을 챙기세요!")
else:
    print("🧥 매우 추운 날씨입니다. 두껍게 입으세요!")
````

````python
## 💻 예제 5 — 양수 / 음수 / 영 판별기
## 생각해보기: `elif` 없이 `if`만 3개 써도 될까? 어떤 차이가 있을까?

num = int(input("숫자를 입력하세요: "))

if num > 0:
    print("양수입니다.")
elif num < 0:
    print("음수입니다.")
else:
    print("0입니다.")
````

## 핵심 개념

<span style="position:left;display:inline-block;">

| 개념   | 설명             | 예시                          |
| ---- | -------------- | --------------------------- |
| elif | 여러 조건 추가       | `elif score >= 80:`         |
| and  | 모두 참일 때 True   | `height >= 140 and age >= 12` |
| or   | 하나라도 참이면 True  | `face or password`          |
| not  | 참/거짓 반전        | `not is_raining`            |

</span>

````python
## 💻 예제 1 — 점수 등급 프로그램


score = int(input("점수 입력: "))

if score >= 90:
    print("A등급 — 최우수!")
elif score >= 80:
    print("B등급 — 우수!")
elif score >= 70:
    print("C등급 — 보통")
elif score >= 60:
    print("D등급 — 노력 필요")
else:
    print("F — 재시험 대상")
````

````python
## 💻 예제 2 — 놀이공원 입장 조건 (and)

height = int(input("키 입력(cm): "))
age = int(input("나이 입력: "))

if height >= 140 and age >= 12:
    print("🎢 탑승 가능!")
else:
    print("🚫 탑승 불가 — 키 140cm 이상, 만 12세 이상이어야 합니다.")
````

````python
## 💻 예제 3 — 스마트폰 잠금 해제 (or)


face = True
password = False

if face or password:
    print("🔓 잠금 해제!")
else:
    print("🔒 접근 불가")
````

````python
## 💻 예제 4 — not 연산자 활용

logged_in = False

if not logged_in:
    print("로그인이 필요합니다.")
else:
    print("환영합니다!")
````

````python
## 💻 예제 5 — 중첩 조건문 (Nested if) — AI 음식 추천

weather = input("날씨 입력 (맑음/비): ")
money = int(input("가진 돈 입력(원): "))

if weather == "비":
    if money >= 10000:
        print("🍲 따뜻한 국밥 추천!")
    elif money >= 5000:
        print("🍜 라면 추천!")
    else:
        print("🏠 집에 있는 재료로 요리하세요!")
else:
    if money >= 10000:
        print("🍔 햄버거 추천!")
    elif money >= 5000:
        print("🍕 편의점 도시락 추천!")
    else:
        print("🏃 집에 있는 과자 드세요!")
````

````python
## 💻 예제 6 — 자율주행 판단 시스템

speed = int(input("현재 속도(km/h): "))
obstacle = input("장애물 감지? (yes/no): ")
battery = int(input("배터리 잔량(%): "))

if battery < 10:
    print("⚠️ 배터리 부족! 즉시 충전소로 이동합니다.")
elif obstacle == "yes" and speed > 30:
    print("🛑 장애물 감지! 급감속합니다.")
elif obstacle == "yes":
    print("⚠️ 장애물 감지! 속도를 줄입니다.")
elif speed > 100:
    print("🚨 과속! 속도를 줄입니다.")
else:
    print("✅ 정상 주행 중입니다.")
````

<h2>RPG 전투 게임 제작</h2>
</div>

* if/elif/else
* 랜덤 공격
* 회복 조건
* 몬스터 AI판단

````python
import random

# =========================
# 플레이어와 몬스터 설정
# =========================

player_hp = 100
player_mp = 50
monster_hp = 100

print("🐉 몬스터 전투 시작!")

# =========================
# 게임 반복
# =========================

while player_hp > 0 and monster_hp > 0:

    print("\n===================")
    print(f"🧑 플레이어 HP : {player_hp}  MP : {player_mp}")
    print(f"👾 몬스터 HP : {monster_hp}")

    # -------------------------
    # 플레이어 행동 선택
    # -------------------------
    print("\n1. 일반 공격")
    print("2. 강한 공격 (30% 실패 확률)")
    print("3. 마법 공격 (MP 20 필요)")
    print("4. 회복")

    choice = input("행동 선택 : ")

    # =========================
    # 플레이어 조건문
    # =========================

    if choice == "1":
        damage = random.randint(10, 20)
        monster_hp -= damage
        print(f"\n⚔ 일반 공격!")
        print(f"{damage} 데미지!")

    elif choice == "2":
        miss = random.randint(1, 100)
        if miss <= 30:
            print("\n💨 강공격 실패!")
        else:
            damage = random.randint(20, 35)
            monster_hp -= damage
            print(f"\n🔥 강공격 성공!")
            print(f"{damage} 데미지!")

    elif choice == "3":
        if player_mp >= 20:
            damage = random.randint(25, 40)
            monster_hp -= damage
            player_mp -= 20
            print(f"\n✨ 마법 공격!")
            print(f"{damage} 데미지! (MP -20)")
        else:
            print("\n❌ MP가 부족합니다!")

    elif choice == "4":
        heal = random.randint(15, 25)
        player_hp += heal
        player_mp += 10
        print(f"\n💚 HP 회복! +{heal}")
        print(f"🔵 MP 회복! +10")

    else:
        print("\n❌ 잘못 입력!")

    # 몬스터 사망 확인
    if monster_hp <= 0:
        print("\n🎉 몬스터를 물리쳤다!")
        break

    # =========================
    # 몬스터 AI (조건문 활용)
    # =========================

    print("\n👾 몬스터 차례!")

    if monster_hp < 30:
        # 체력이 낮으면 회복
        heal = random.randint(10, 20)
        monster_hp += heal
        print(f"👾 몬스터 회복! +{heal}")

    elif player_hp < 40:
        # 플레이어 체력이 낮으면 강공격
        damage = random.randint(20, 30)
        player_hp -= damage
        print(f"👾 몬스터 필살기! {damage} 데미지!")

    else:
        # 일반 공격
        damage = random.randint(10, 20)
        player_hp -= damage
        print(f"👾 몬스터 공격! {damage} 데미지!")

    # 플레이어 사망 확인
    if player_hp <= 0:
        print("\n💀 플레이어 패배...")
        break

print("\n🎮 게임 종료")
````
